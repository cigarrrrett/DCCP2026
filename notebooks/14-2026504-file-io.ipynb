```python
text = "the cat sat on the mat the cat slept"
words = text.split()
print(words)
freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1
with open('freq.txt', 'w', encoding='utf-8') as f:
    for (word, count) in sorted(freq.items()):
        f.write(f'{word}\t{count}\n')
```
