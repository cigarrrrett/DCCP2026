```python
import numpy as np
import pandas as pd

works = pd.DataFrame({
    '작가': ['김유정', '김유정', '김유정',
    '현진건', '현진건',
    '이상', '이상',
    '나도향', '나도향',
    '채만식', '이효석'],
    '작품': ['봄봄', '동백꽃', '만무방',
    '운수 좋은 날', 'B사감과 러브레터',
    '날개', '권태',
    '벙어리 삼룡이', '물레방아',
    '레디메이드 인생', '메밀꽃 필 무렵'],
    '연도': [1935, 1936, 1934,
    1924, 1925,
    1936, 1937,
    1925, 1925,
    1934, np.nan], # 이효석 '메밀꽃 필 무렵' 발표연도가 누락됨
    '글자수': [12500, 9800, 13600,
    11200, 7400,
    18300, 14100,
    16800, 14500,
    22000, 11800],
})

authors = pd.DataFrame({
    '작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
    '생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
    '몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})
```

```python
print(works.shape) 
print(works.dtypes) 
print(works.describe()) 
```

NaN은 정수형에는 없는 표현이기에 NaN이 섞이면 정수열도 float로 표현된다. 

```python
works.isna().sum()
```

연도 열에 한 건의 결측지가 있다. 

```python
works2=works.fillna({'연도':1936})
works2['연도']=works2['연도'].astype('int')
print(works2.dtypes)
print(works2.tail(3))
```

```python
after_1930=works2[works2['연도']>=1930]
kim_or_lee=works2[works2['작가'].isin(['김유정', '이상'])]
print(after_1930)
print(kim_or_lee)
```

```python
def categorize(n: int) -> str:
    if n < 10000: 
        return "짧음"
    elif n > 15000: 
        return "긴"
    else:
        return "보통"

works2['분량']=works2['글자수'].apply(categorize)
print(works2[["작품", "글자수", "분량"]])
```

```python
print(works2.sort_values(["연도", "글자수"], ascending=[True, False]))
```

```python
print(works2.groupby('작가')['글자수'].agg(['mean', 'count', 'sum']))
```

```python
print(works2['분량'].value_counts()) 
```

```python
print(works2.groupby(["작가", "분량"]).size())
```

가장 많은 작품 수를 가진 (작가, 분량) 조합은 (김유정, 보통)이고 이 내용은 김유정 작가가 10000자 이상 15000자 이하의 중편 작품을 선호했음을 알려준다. 

```python
full=pd.merge(works2, authors, on="작가", how="left")
print(full)
```

left는 works2를 기준으로 병합하기 떄문에 authors에는 있지만 works2에는 없는 작가는 포함되지 않는다. 

```python
full['집필 당시 나이'] = full['연도'] - full['생년']
print(full[["작가", "작품", "연도", "생년", "집필 당시 나이"]])
```

```python
result = full.groupby('작가')['집필 당시 나이'].mean().sort_values(ascending=True)
print(result)
```

나도향이 가장 어린나이(평균 23.5세)에 작품을 집필했고, 채만식이 가장 많은 나이(평균32세)에 작품을 집필했다.
